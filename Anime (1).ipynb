```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("anime.csv")
df.head()
```


Make a new column for episode count


```python
df["Episode_Count"] = df["Title"].str.extract(r'(\d+) eps')
df.head()
```


Make a new column For time stamp


```python
df["Timestamp"] = df["Title"].str.extract(r'([A-Z][a-z]{2} \d{4} - [A-Z][a-z]{2} \d{4})')
df.head()
```


Which anime has the highest score


```python
df[df["Score"] == df["Score"].max()]
```


Top 5 highest scoring anime


```python
df.sort_values(by="Score", ascending=False).head(5)
```


Which anime has the highest episode count


```python
df[df["Episode_Count"] == df["Episode_Count"].max()]
```


Animes with top 5 episode count


```python
df.sort_values(by="Episode_Count", ascending=False).head(5)
```


The longest running anime


```python
df["Start_Date"] = pd.to_datetime(df["Timestamp"].str.split(" - ").str[0],format="%b %Y")

df["End_Date"] = pd.to_datetime(df["Timestamp"].str.split(" - ").str[1],format="%b %Y")

df["Duration_Months"] = ((df["End_Date"].dt.year - df["Start_Date"].dt.year) * 12+ (df["End_Date"].dt.month - df["Start_Date"].dt.month))

df.nlargest(1, "Duration_Months")[["Title", "Timestamp", "Duration_Months"]]
```
